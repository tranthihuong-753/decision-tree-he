```python
import numpy as np
from scipy.integrate import quad
```

# Bậc 8 

```python
# Thiết lập tham số
N = 8  # Bậc đa thức
a = -2.0  # Giới hạn tích phân dưới
b_val = 2.0   # Giới hạn tích phân trên
epsilon = 0.2 # Kích thước cửa sổ bỏ qua [-epsilon, epsilon]
```

```python
# Hàm tính tích phân cho M_ij
def integral_M(k):
    # Tích phân của x^k * w(x) trên [-2, 2]
    # Do w(x)=1 trên [-2, -0.2] U [0.2, 2]

    if k == -1: # Trường hợp đặc biệt (không xảy ra ở đây)
        return 0.0

    # Hàm nguyên hàm của x^k là x^(k+1) / (k+1)
    def anti_derivative(x, k):
        return x**(k + 1) / (k + 1)

    # Tính tích phân trên hai khoảng
    int_left = anti_derivative(-epsilon, k) - anti_derivative(a, k)
    int_right = anti_derivative(b_val, k) - anti_derivative(epsilon, k)

    return int_left + int_right

# Xây dựng Ma trận M
M = np.zeros((N + 1, N + 1))
for i in range(N + 1):
    for j in range(N + 1):
        k = i + j
        M[i, j] = integral_M(k)

# Tính vector b
# Hàm tính tích phân cho b_j
def integral_b(j):
    # Tích phân của I0(x) * x^j * w(x) trên [-2, 2]
    # I0(x)*w(x) = 1 trên [0.2, 2] và 0 ở các vùng khác

    # Hàm nguyên hàm của x^j là x^(j+1) / (j+1)
    def anti_derivative(x, j):
        return x**(j + 1) / (j + 1)

    # Tính tích phân trên khoảng [epsilon, b]
    return anti_derivative(b_val, j) - anti_derivative(epsilon, j)

# Xây dựng Vector b
b = np.zeros(N + 1)
for j in range(N + 1):
    b[j] = integral_b(j)
```

```python
# Giải hệ phương trình tuyến tính M * c = b
try:
    coefficients = np.linalg.solve(M, b)
    print("Các hệ số đa thức (c0 đến c16) đã tính được:")
    print(coefficients)
except np.linalg.LinAlgError as e:
    print(f"Lỗi giải hệ phương trình: {e}. Ma trận M có thể suy biến (singular).")
```
